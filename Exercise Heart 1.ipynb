## Hjertets anatomi og fysiologi

### Hjertets opbygning og blodets vej

Hjertet består af fire kamre: højre atrium (forkammer), højre ventrikel (hjertekammer), venstre atrium og venstre ventrikel. Blodet cirkulerer i to kredsløb: det lille (pulmonal) kredsløb mellem hjertet og lungerne, og det store (systemiske) kredsløb mellem hjertet og resten af kroppen.

Det iltfattige blod fra kroppen returnerer via vena cava til højre atrium, passerer gennem trikuspidalklappen til højre ventrikel, og pumpes videre gennem pulmonalklappen ud i lungearterierne til lungerne, hvor det iltes. Det iltede blod vender tilbage via lungevenerne til venstre atrium, passerer mitralklappen til venstre ventrikel, og pumpes endelig ud i aorta via aortaklappen.

Hjertets to arbejdsfaser er systole (sammentrækning), hvor ventriklerne kontraherer og pumper blod ud, og diastole (afslapning), hvor ventriklerne fyldes med blod fra atrierne.

### Cardiac output (minutvolumen)

Hjertets minutvolumen (CO) angiver den mængde blod, hjertet pumper ud pr. minut:

$$CO = HR \cdot SV$$

hvor $HR$ er hjertefrekvensen (slag/min) og $SV$ er slagvolumen (ml/slag) — den mængde blod, der pumpes ud ved hvert slag. En voksen i hvile har typisk $HR \approx 70$ slag/min og $SV \approx 70$ mL, hvilket giver $CO \approx 5$ L/min.

### Dilateret kardiomyopati

Ved dilateret kardiomyopati udvider (dilaterer) venstre ventrikel sig, og hjertemusklen svækkes. Det øgede ventrikelvolumen fører til et øget slutdiastolisk volumen (preload), men kontraktiliteten er nedsat, så slagvolumen falder. Som kompensation stiger hjertefrekvensen. Den dilaterede ventrikel øger vægstressen ifølge Laplaces lov (se næste afsnit), da en større radius ved samme tryk kræver større spænding i væggen. Over tid reagerer hjertemusklen på den øgede belastning med hypertrofi (fortykkelse af væggen), som er et forsøg på at normalisere vægstressen — men som selv kan føre til nedsat hjertepumpefunktion.



## Laplaces lov og vægstress

### Laplaces lov for en sfærisk membran

Laplaces lov beskriver sammenhængen mellem trykforskellen over en kugleformet struktur, dens geometri og den mekaniske spænding/stress i væggen. For en tyndvægget kugle ($w \ll r$) gælder:

$$\sigma = \frac{P \cdot r}{2 \cdot w}$$

hvor $\sigma$ er vægstressen, $P$ er trykket inde i kammeret, $r$ er indre radius og $w$ er vægtykkelsen.

### Enheder

Formlen indeholder forholdet $r/w$. Hvis $r$ og $w$ angives i de samme længdeenheder, går de ud med hinanden:

$$\left[\frac{r}{w}\right] = \frac{\text{cm}}{\text{cm}} = 1 \quad \text{(dimensionsløs)}$$

Dermed har $\sigma$ samme enhed som $P$:

$$[\sigma] = [P] \cdot \frac{[r]}{[w]} = \text{mmHg} \cdot 1 = \text{mmHg}$$

Det er praktisk i klinisk sammenhæng, fordi tryk allerede måles i mmHg og radier/vægtykkelse typisk i cm. Notebooken her bruger da også mmHg for vægstress.

> OBS: Hvis man angiver $r$ og $w$ i forskellige enheder (f.eks. $r$ i m og $w$ i cm), skal man omregne, inden man sætter ind i formlen — ellers får man en forkert enhed og forkert talværdi.

### Vægstress afhænger af geometri

Vægstress er den kraft, muskelfibrene skal generere for at opretholde et givet tryk. Det er værd at bemærke:

- Større radius → større vægstress (samme tryk kræver mere kraft)
- Tykkere væg → mindre vægstress (belastningen fordeles over mere væv)
- Højere tryk → større vægstress (logisk nok)

### Klinisk relevans

Ved dilateret kardiomyopati stiger $r$, og $w$ falder typisk (eller i hvert fald holder ikke trit med dilatationen). Begge effekter øger $\sigma$, hvilket forklarer, hvorfor en dilateret ventrikel er mere belastet og med tiden udvikler hypertrofi som kompensation. Myokardie-iltforbruget er tæt knyttet til vægstress × volumen, så øget vægstress bidrager til myokardie-iskæmi.



## Vægstress i aorta og venstre ventrikel

Når man sammenligner vægstressen i aorta (AO) og venstre ventrikel (LV) gennem en hjertecyklus, ser man en tydelig forskel — og det er ikke tilfældigt.

Aorta er et rør med en relativ stor radius (ca. 1,25 cm) og en tynd væg (ca. 0,2 cm), så forholdet $r/w$ er stort. Aortaklappen er lukket i diastolen, og aorta holder trykket oppe på 80–90 mmHg, så aortavæggen er spændt gennem hele cyklussen. Det afspejler, at aorta fungerer som en trykbuffer (Windkessel-effekt).

Venstre ventrikel har til sammenligning en tykkere muskelvæg (ca. 1 cm), så $r/w$ er mindre. I diastolen er trykket i LV lavt (10–20 mmHg), og vægstressen er derfor lavere end i aorta, selvom volumenet (og radius) er størst. Når ventriklen kontraherer, stiger trykket meget mere, end radius falder, så vægstressen i LV stiger kraftigt gennem den isovolumetriske kontraktion og topper omkring peak systole. Her er LV-stressen langt højere end aortas.

Belastningen ligger altså primært i venstre ventrikel under systolen og i aorta som en vedvarende mekanisk belastning gennem hele cyklussen.

## Exercise: Beregn vægstress i venstre ventrikel

**Baggrund:**  
Laplaces lov for en kugleformet struktur kan skrives som:
$$
P = \frac{2 \cdot \sigma \cdot w}{r} = \frac{2 \cdot T}{r}
$$
Laplaces lov for en cylinderformet struktur kan skrives som:
$$
P=\frac{T}{r}
$$
hvor:
*   ($P$) = indre tryk (mmHg)
*   ($\sigma$) = vægstress (mmHg, når $r$ og $w$ har samme enhed)
*   ($T$) = væg tension (mmHg·cm)
*   ($w$) = vægtykkelse (cm)
*   ($r$) = indre radius (cm)

**Opgave:**
Antag at venstre ventrikel har:
* Radius ($r = 3 \text{ cm}$)
* Vægtykkelse ($w = 1 \text{ cm}$)
* Tryk ($P = 120 \text{mmHg}$)

1.  Lav en vægstress ($\sigma$) model i funktionen `calculate_wallstress` for en sphere i en funktion (Lav en funktion som tager de nødvendige inputs, og returnerer vægspændingen).



```python
import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(precision=3, suppress=True)

# Answer
def calculate_wallstress(r, w, P):
    wallstress = P * r / w / 2.0
    return wallstress

print(f"Vægstress ved r=3 cm, w=1 cm, P=120 mmHg: {calculate_wallstress(3.0, 1.0, 120.0):.0f} mmHg")

Rratio = {"Normal": 1.0}
# radius from 1.0 to 6 in 0.1 steps
R = np.array(range(10, 61)) * 0.1
P = 120.0
w = 1.0
for name,r_ratio in Rratio.items():
    R_modified = R * r_ratio
    stress = calculate_wallstress(R_modified, w, P)
    plt.plot(R, stress)

plt.title("Wall Stress vs Radius")
plt.xlabel("Radius (cm)")
plt.ylabel("Wall Stress (mmHg)")
plt.show()
```
